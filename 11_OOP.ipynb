##  Inheritance
- **Inheritance** - Child(derived) class inherits or derives the properties & methods from parent(base) class
- The name of the parent class is passed inside parentheses when defining the class class
- there are three different types of inheritance: Single, Multi-level, multiple inheritance

- **Single inheritance** - occurs when a child class inherits methods & attr from only one parent class(superclass or base class)
-It is simplest, most linear form of class
- *benefits*
- prenvents code duplication
- inheritance path is a straight line, making code execution easy to trace 
- No Ambiguity

```python
#single inheritance
class Car:
    @staticmethod
    def start():
        print ("car started")

    @staticmethod
    def stop():
        print ("car stopped")

class ToyotaCar(Car): #here we are inheriting the Car class
    def __init__(self, name,color):
        self.name = name
        self.color = color

car1 = ToyotaCar("Fortuner","black")
car2 = ToyotaCar("prius","yellow")

print("Car 1 Name:",car1.name)
car1.start()
print("Car 1 Color:", car1.color)
print("Car 2 name is", car2.name, ", It's color is", car2.color)
car2.start()
```

**Multi-level inheritance**
- occursss when a child class inherits from a parent class, which in turns inherits from another parent class
- It is linear, multi-generational chain of classes where features trickles down from top to bottom
- There can be incountable no of generations in multi level inheritance
- *benefits*
- Granular Organization: allows you to breake down code into clear, logical steps form general to specific 
- Cumulative Reusability: the lowest class automatically gets combined logic of all classes above it 
- clean Lineage: Avoids the structural messiness of multiple inheritance because the lookup path remains a straight line

```python
#Multi- level inheritance
class Car:                          #1st level parent class(Grandparent class)
    @staticmethod
    def start():
        print ("car started")

    @staticmethod
    def stop():
        print ("car stopped")

class ToyotaCar(Car):               #2nd level child class, inheriting from Car class (Parent class)
    def __init__(self,brand):
        self.brand = brand

class Fortuner(ToyotaCar):          #3rd level child class, inheriting from Car class & ToyotaCar class (Grandchild class)
    def __init__(self,type):
        self.type = type

car1 = Fortuner("diesel")
car1.start()
```

**Multiple Inheritance**
- a feature that allows a child class to inherit attributes & methods from more than one parent class
- this enables developers to combine the functionality of multiple independent classes into a single, cohesive subclass
- *Benefits*
- code reusability: you can write a distinct feature once and apply it ti completely different types of objects 
- Modular component Design: it allows you to built classes using a plug-&-play mindset
- Accurate Real-world Modeling: Many complex real-world objects naturally posses multiple distinct identities or capabilities

```python
#multiple inheritance
class Car:
    @staticmethod
    def start():
        print ("car started")

    @staticmethod
    def stop():
        print ("car stopped")

class ElectricSystem:
    @staticmethod
    def battery():
        print ("battery is charged")

class HybridCar(Car, ElectricSystem): #Here we are inheriting from both Car class and ElectricSystem class
    def __init__(self, name, color):
        self.name = name
        self.color = color

car1 = HybridCar("Prius","white")
car2 = HybridCar("Fortuner","black")

print("Car 1 Name:",car1.name)
print("Car 1 Color:",car1.color)
car1.start()
car1.battery()

print("\n","Car 2 Name is",car2.name,"It's Color is",car2.color)
car2.start()
car2.battery()
```

**Super Method**
- super()method is used to access methods & call the constructor of the parent(super) class in the child class.



```python
#super method
class Car:
    def __init__(self, type):
        self.type = type

    @staticmethod
    def start():
        print ("car started")

    @staticmethod
    def stop():
        print ("car stopped")

class ToyotaCar(Car):
    def __init__(self, name, type):
        super().__init__(type) #super() function is used to give access to the methods of a parent class from the child class.
        self.name = name
        super().start() #calling the start method of parent class

car1 = ToyotaCar("Fortuner","diesel")
print(car1.type)
```

**Class method**
- A class method is bound to the class & not the object of the class
- It recieves the class as an implicit first argument.


```python
#class method
class Person:
    name = "Anonymous" #class attribute

    # def changeName(self, name): #instance method
    #     self.name = name #normal object attribute using this we can change the name of the object but not the class attribute
    #     Person.name = name #Method 1 to change class attribute
    #     self.__class__.name = name #Method 2 to change class attribute

    @classmethod
    def changeName(cls, name): #class method
        cls.name = name #using class method we can change the class attribute
        
p1 = Person()
p1.changeName("John")
print(p1.name)
print(Person.name) #class attribute can be accessed using class name
```

**Property Decorator**
- we use @property decorator on any method in the class to use the method as a property.

```python
#without Using property decorator
class Student:
    def __init__(self, phy, chem, maths):
        self.phy = phy
        self.chem = chem
        self.maths = maths

        self.percentage = str((self.phy + self.chem + self.maths)/3) + "%" #it's optional to write self.phy + self.chem + self.maths we can write phy + chem + maths too or we ca write mix of both. all will give same result
        #it's optional to convert the % into str & add % sign at the end. If we don't convert it into str then it will give error while adding % sign at the end.
    def calcper(self): #adding this will update the % but using this method we are doing double work
        self.percentage = str((self.phy+ self.chem + self.maths)/3) +"%"
    
s1 = Student(90, 80, 92)
print("Percentage of Student 1 is:", s1.percentage)

s1.phy= 86
print(s1.phy) #here marks will be updated 
s1.calcper()
print(s1.percentage) # but when we try to get the percentage it will not be updated until we add s1.calcper()
```

```python
#Using property decorator
class Student:
    def __init__(self, phy, chem, maths):
        self.phy = phy
        self.chem = chem
        self.maths = maths

    @property
    def percentage(self):
        return str((self.phy + self.chem + self.maths) / 3) +"%"
s1 = Student(90, 80, 92)
print("Percentage of Student 1 is:", s1.percentage)

s1.phy= 86
print(s1.percentage)
```

**Operators & Dunder functions**
- Operators in Python can be overloaded using dunder methods. 
- These methods are called when a given operator is used on the objects. 
- Operators in Python can be overloaded using the following methods: 
- p1+p2      # Addition p1.__ add__(p2) 
- p1-p2      # Subtraction p1.__ sub__(p2) 
- p1*p2      # Multiplication p1.__ mul__(p2) 
- p1/p2      # Division p1.__ truediv__(p2) 
- p1//p2     # Addition p1.__ floordiv__(p2) 
- p1>p2      # Greater then p1.__ gt__(p2)
- str__()    # used to set what gets displayed upon calling str(obj) 
- __ len__() # used to set what gets displayed upon calling.__len__() or 
len(obj) 

```python
#operators and dunder functions
class Complex:
    def __init__(self, real, img):
        self.real = real
        self.img = img

    def showNo(self):
        print(self.real,"i+", self.img,"j")

    def __add__(self, num2):
        newReal = self.real + num2.real
        newImg = self.img + num2.img
        return Complex(newReal, newImg)

num1 = Complex(1,3)
num1.showNo()

num2 = Complex(4,6)
num2.showNo()

# num3 = num1.add(num2) #when you are not Using Dunder Function
num3 = num1 + num2 #when using Dunder Function
num3.showNo()

```

##  Encapsulation 
- **Encapsulation** - Hiding data inside methods/classes (Wrapping data and functions into a single unit (object))

```python
# Encapsulation (Create Account class with 2 attributes: balance & account no.)
class Account:
    def __init__(self, bal, acc):
        self.balance = bal
        self.account_no = acc

    def debit(self, amount):
        self.balance -= amount
        print("Rs.", amount , "was debited")
        print("total balance =", self.get_balance())

    def credit(self, amount):
        self.balance += amount
        print("Rs.", amount , "was credited") 
        print("total balance =", self.get_balance())

    def get_balance(self):
        return self.balance

acc1 = Account (10000, 12345)
acc1.debit(1000)
acc1.credit(500)
acc1.credit(500000)
acc1.debit(700)


```

**Private(like) attributes & methods**
- conceptional implementation in python
- private attributes & methods are meant to be used within class and are not accessible from outside the class
- we used double underscore(__) before that attribute
- Therefore we will not be able to access it outside the class
- basically done to prevent exposing are instance attr outside the class

```python
#Private(like) attributes & methods
class Account:
    def __init__(self,acc_no, acc_pass):
        self.acc_no = acc_no
        self.__acc_pass = acc_pass

    def reset_pass(self):
        print(self.__acc_pass) #we can access private variable inside the class using method of the class 

acc1 = Account("12345","Anhum123")
print(acc1.acc_no)
# print(acc1.__acc_pass) if we will try to print acc_pass it will give error because it is private variable and can not be accessed outside the class
print(acc1.reset_pass())
```
