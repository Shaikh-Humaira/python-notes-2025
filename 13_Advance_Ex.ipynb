Q2. Write a program to print third, fifth and seventh element from a list using enumerate 
function.

```python
#Q2.
no = [1, 2, 3, 4, 5, 6, 7, 8]

for index, item in enumerate(no):
    if index == 2 or index == 4 or index == 6:
        print(item)
```

Q3. Write a list comprehension to print a list which contains the multiplication table of a 
user entered number.

```python
#Q3.
print ("To get the Multiplication table")

no = int(input("Enter the number:"))

table = [no*i for i in range(1, 11)]
print(f"Table of {no}:{str(table)} ")
```

Q8. Write a program to filter a list of numbers which are divisible by 5.

```python
a = [1,75,3,65,37,346,27,25,753,26,35,68,393,940,5856,35780,56,9890,69,68, 98,65,457]
def mul5(n):
    if (n%5 == 0):
        return True
    return False
f = list(filter(mul5, a))
print(f)
```

Q9. Write a program to find the maximum of the numbers in a list using the reduce 
function. 

```python
from functools import reduce
l = [1,75,3,65,37,346,27,25,753,26,35,68,393,940,5856,35780,56,9890,69,68, 98,65,457]
def greater(a,b):
    if (a>b):
        return a
    return b
print("Greatest no from the list is:",reduce(greater,l))
```
